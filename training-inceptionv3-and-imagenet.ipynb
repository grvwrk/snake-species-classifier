```python
import torch
from torch.utils.data import Dataset, DataLoader
from PIL import Image
from pathlib import Path
from torchvision.models import Inception_V3_Weights

weights = Inception_V3_Weights.DEFAULT

# Map actual class IDs to contiguous labels: 0, 1, 2, ...
class_ids = sorted(train_df["class_id"].unique())
class_to_idx = {class_id: i for i, class_id in enumerate(class_ids)}

idx_to_species = (
    train_df.groupby("class_id")["binomial"]
    .first()
    .to_dict()
)

NUM_CLASSES = len(class_to_idx)

print("Number of classes:", NUM_CLASSES)
print("Example mapping:", list(class_to_idx.items())[:5])
print("Example species:", list(idx_to_species.items())[:5])
```

```python
from torchvision import transforms

train_transform = transforms.Compose([
    transforms.RandomResizedCrop(299),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

test_transform = weights.transforms()


class SnakeDataset(Dataset):
    def __init__(self, dataframe, image_dir, class_to_idx, transform=None):
        self.df = dataframe.reset_index(drop=True)
        self.image_dir = Path(image_dir)
        self.class_to_idx = class_to_idx
        self.transform = transform

        self.image_paths = {
            p.stem: p
            for p in self.image_dir.rglob("*")
            if p.is_file()
        }

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        image_path = self.image_paths[str(row["UUID"]).strip()]

        image = Image.open(image_path).convert("RGB")

        if self.transform:
            image = self.transform(image)

        label = self.class_to_idx[row["class_id"]]
        return image, label
```

```python
train_dataset = SnakeDataset(
    train_df, TRAIN_DIR, class_to_idx, train_transform
)

test_dataset = SnakeDataset(
    test_df, TEST_DIR, class_to_idx, test_transform
)

train_loader = DataLoader(
    train_dataset,
    batch_size=16,
    shuffle=True,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)

test_loader = DataLoader(
    test_dataset,
    batch_size=16,
    shuffle=False,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)

images, labels = next(iter(train_loader))

print("Images:", images.shape)
print("Labels:", labels.shape)
print("Number of training images:", len(train_dataset))
print("Number of test images:", len(test_dataset))
```

```python

# ============================================================
# Inception V3: Snake Species Classification
# Environment: Modal | NVIDIA L40S | PyTorch 2.8+
# Phase: Classifier-head training
# Precision: BF16 mixed precision
# Checkpoint: Verified, atomic local save
# ============================================================

import os
import gc
from pathlib import Path

import torch
import torch.nn as nn
import torch.optim as optim
from torchvision.models import inception_v3, Inception_V3_Weights
from tqdm.auto import tqdm


# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

EPOCHS = 5
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4

CHECKPOINT_DIR = Path("/root/snake_checkpoints")
CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

BEST_PATH = CHECKPOINT_DIR / "inception_v3_snakes_head.pth"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

if device.type != "cuda":
    raise RuntimeError("CUDA GPU not detected. Check your Modal GPU configuration.")

if not torch.cuda.is_bf16_supported():
    raise RuntimeError("This script requires GPU BF16 support.")

print(f"GPU: {torch.cuda.get_device_name(0)}")
print(f"PyTorch: {torch.__version__}")
print(f"BF16 supported: {torch.cuda.is_bf16_supported()}")
print(f"Checkpoint destination: {BEST_PATH}")

# Performance settings
torch.backends.cudnn.benchmark = True


# ------------------------------------------------------------
# 2. Validate existing notebook variables
# ------------------------------------------------------------

required_variables = [
    "train_loader",
    "val_loader",
    "class_to_idx",
    "idx_to_species",
    "NUM_CLASSES",
]

missing = [
    name for name in required_variables
    if name not in globals()
]

if missing:
    raise RuntimeError(
        f"Missing notebook variables: {missing}. "
        "Run your dataset and DataLoader setup cells first."
    )

if len(class_to_idx) != NUM_CLASSES:
    raise ValueError(
        f"class_to_idx contains {len(class_to_idx)} classes, "
        f"but NUM_CLASSES={NUM_CLASSES}."
    )


# ------------------------------------------------------------
# 3. Initialize pretrained Inception V3
# ------------------------------------------------------------

print("\nInitializing pretrained Inception V3...")

model = inception_v3(weights=Inception_V3_Weights.DEFAULT)

# Disable the auxiliary classifier after loading pretrained weights.
model.aux_logits = False
model.AuxLogits = None

# Replace the final classification layer.
in_features = model.fc.in_features
model.fc = nn.Linear(in_features, NUM_CLASSES)

# Freeze the pretrained backbone.
for parameter in model.parameters():
    parameter.requires_grad = False

# Train only the new classification head.
for parameter in model.fc.parameters():
    parameter.requires_grad = True

# Channels-last can improve convolution performance on NVIDIA GPUs.
model = model.to(
    device=device,
    memory_format=torch.channels_last
)

criterion = nn.CrossEntropyLoss()

optimizer = optim.AdamW(
    model.fc.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

print(f"Number of classes: {NUM_CLASSES}")
print(f"Trainable parameters: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}")
print(f"Training batches: {len(train_loader)}")
print(f"Validation batches: {len(val_loader)}")


# ------------------------------------------------------------
# 4. Safe checkpoint saving
# ------------------------------------------------------------

def save_checkpoint_safely(payload, destination):
    destination = Path(destination)
    destination.parent.mkdir(parents=True, exist_ok=True)

    temp_path = destination.with_name(destination.name + ".tmp")

    try:
        # Normalize metadata to safe, standard Python types.
        safe_payload = dict(payload)

        safe_payload["class_to_idx"] = {
            str(k): int(v)
            for k, v in safe_payload["class_to_idx"].items()
        }

        safe_payload["idx_to_species"] = {
            int(k): str(v)
            for k, v in safe_payload["idx_to_species"].items()
        }

        safe_payload["num_classes"] = int(
            safe_payload["num_classes"]
        )
        safe_payload["best_val_acc"] = float(
            safe_payload["best_val_acc"]
        )
        safe_payload["epoch"] = int(safe_payload["epoch"])

        # Ensure checkpoint tensors reside on CPU.
        safe_payload["model_state_dict"] = {
            key: tensor.detach().cpu().contiguous().clone()
            for key, tensor in safe_payload["model_state_dict"].items()
        }

        # Write to a temporary file first.
        with open(temp_path, "wb") as f:
            torch.save(safe_payload, f)
            f.flush()
            os.fsync(f.fileno())

        # Verify that the checkpoint is readable with restricted loading.
        verified = torch.load(
            temp_path,
            map_location="cpu",
            weights_only=True
        )

        assert "model_state_dict" in verified
        assert len(verified["model_state_dict"]) > 0
        assert "best_val_acc" in verified
        assert "class_to_idx" in verified

        # Replace the previous checkpoint only after verification.
        os.replace(temp_path, destination)

        print(
            f"Verified checkpoint saved: {destination} "
            f"({destination.stat().st_size / 1e6:.1f} MB)"
        )

    except Exception:
        if temp_path.exists():
            temp_path.unlink()
        raise


# ------------------------------------------------------------
# 5. Initialize best-model tracking
# ------------------------------------------------------------

best_val_acc = -1.0
best_epoch = 0

# Save an initial recovery checkpoint before training begins.
# This protects the initialized model if the first epoch fails.
initial_state = {
    key: tensor.detach().cpu().clone()
    for key, tensor in model.state_dict().items()
}

initial_payload = {
    "model_state_dict": initial_state,
    "class_to_idx": class_to_idx,
    "idx_to_species": idx_to_species,
    "num_classes": NUM_CLASSES,
    "best_val_acc": -1.0,
    "epoch": 0,
    "architecture": "inception_v3",
    "phase": "classifier_head",
}

save_checkpoint_safely(initial_payload, BEST_PATH)

del initial_state, initial_payload
gc.collect()


# ------------------------------------------------------------
# 6. Training and validation
# ------------------------------------------------------------

history = []

for epoch in range(EPOCHS):

    # --------------------------------------------------------
    # Training
    # --------------------------------------------------------

    # Keep the frozen backbone in evaluation mode so BatchNorm
    # running statistics do not drift during head-only training.
    model.eval()
    model.fc.train()

    train_loss_sum = 0.0
    train_correct = 0
    train_total = 0

    train_bar = tqdm(
        train_loader,
        desc=f"Epoch {epoch + 1}/{EPOCHS}"
    )

    for images, labels in train_bar:
        images = images.to(
            device,
            non_blocking=True,
            memory_format=torch.channels_last
        )
        labels = labels.to(device, non_blocking=True).long()

        optimizer.zero_grad(set_to_none=True)

        with torch.autocast(
            device_type="cuda",
            dtype=torch.bfloat16
        ):
            outputs = model(images)
            loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        batch_size = labels.size(0)
        train_loss_sum += loss.item() * batch_size

        predictions = outputs.argmax(dim=1)
        train_correct += (predictions == labels).sum().item()
        train_total += batch_size

        train_bar.set_postfix(
            loss=f"{loss.item():.4f}",
            acc=f"{100 * train_correct / train_total:.2f}%"
        )

    train_loss = train_loss_sum / train_total
    train_acc = 100.0 * train_correct / train_total

    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    model.eval()

    val_loss_sum = 0.0
    val_correct = 0
    val_total = 0

    with torch.inference_mode():
        for images, labels in tqdm(
            val_loader,
            desc="Validation",
            leave=False
        ):
            images = images.to(
                device,
                non_blocking=True,
                memory_format=torch.channels_last
            )
            labels = labels.to(device, non_blocking=True).long()

            with torch.autocast(
                device_type="cuda",
                dtype=torch.bfloat16
            ):
                outputs = model(images)
                loss = criterion(outputs, labels)

            batch_size = labels.size(0)
            val_loss_sum += loss.item() * batch_size

            predictions = outputs.argmax(dim=1)
            val_correct += (predictions == labels).sum().item()
            val_total += batch_size

    val_loss = val_loss_sum / val_total
    val_acc = 100.0 * val_correct / val_total

    print(f"\nEpoch {epoch + 1}/{EPOCHS}")
    print(
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.2f}%"
    )
    print(
        f"Val Loss:   {val_loss:.4f} | "
        f"Val Acc:   {val_acc:.2f}%"
    )

    history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_acc": train_acc,
        "val_loss": val_loss,
        "val_acc": val_acc,
    })

    # --------------------------------------------------------
    # Save only when validation accuracy improves
    # --------------------------------------------------------

    if val_acc > best_val_acc:

        candidate_state = {
            key: tensor.detach().cpu().clone()
            for key, tensor in model.state_dict().items()
        }

        payload = {
            "model_state_dict": candidate_state,
            "class_to_idx": class_to_idx,
            "idx_to_species": idx_to_species,
            "num_classes": NUM_CLASSES,
            "best_val_acc": float(val_acc),
            "epoch": epoch + 1,
            "architecture": "inception_v3",
            "phase": "classifier_head",
        }

        # Update the file before updating in-memory best metrics.
        save_checkpoint_safely(payload, BEST_PATH)

        best_val_acc = float(val_acc)
        best_epoch = epoch + 1

        print(
            f"New best validation accuracy: {best_val_acc:.2f}% "
            f"(epoch {best_epoch})"
        )

        del candidate_state, payload

    print("-" * 65)


# ------------------------------------------------------------
# 7. Restore the best saved model
# ------------------------------------------------------------

print("\nLoading best verified checkpoint...")

checkpoint = torch.load(
    BEST_PATH,
    map_location="cpu",
    weights_only=True
)

model.load_state_dict(checkpoint["model_state_dict"])
model.to(device, memory_format=torch.channels_last)
model.eval()

print("\nTraining complete.")
print(f"Best epoch: {checkpoint['epoch']}")
print(f"Best validation accuracy: {checkpoint['best_val_acc']:.2f}%")
print(f"Checkpoint: {BEST_PATH}")
print(f"Checkpoint size: {BEST_PATH.stat().st_size / 1e6:.1f} MB")


# ------------------------------------------------------------
# 8. Optional: download the checkpoint from the notebook
# ------------------------------------------------------------

print(
    "\nIMPORTANT: This path is local to the Modal environment. "
    "Download the checkpoint before the environment is terminated."
)
```

```python

# ============================================================
# Inception V3: Fine-tuning final Inception blocks
# Modal | NVIDIA L40S | PyTorch 2.8+
# ============================================================

import os
import gc
from pathlib import Path

import torch
import torch.nn as nn
import torch.optim as optim
from torchvision.models import inception_v3, Inception_V3_Weights
from tqdm.auto import tqdm


# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

EPOCHS = 15
BACKBONE_LR = 1e-5
CLASSIFIER_LR = 1e-4
WEIGHT_DECAY = 1e-4

CHECKPOINT_DIR = Path("/root/snake_checkpoints")
CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

HEAD_CHECKPOINT = CHECKPOINT_DIR / "inception_v3_snakes_head.pth"
BEST_PATH = CHECKPOINT_DIR / "inception_v3_snakes_finetuned_best.pth"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

if device.type != "cuda":
    raise RuntimeError("CUDA GPU not detected.")

if not torch.cuda.is_bf16_supported():
    raise RuntimeError("This script requires BF16 support.")

print(f"GPU: {torch.cuda.get_device_name(0)}")
print(f"PyTorch: {torch.__version__}")
print(f"BF16 supported: {torch.cuda.is_bf16_supported()}")
print(f"Loading checkpoint: {HEAD_CHECKPOINT}")


# ------------------------------------------------------------
# 2. Validate dataset variables
# ------------------------------------------------------------

required_variables = [
    "train_loader",
    "val_loader",
    "class_to_idx",
    "idx_to_species",
    "NUM_CLASSES",
]

missing = [name for name in required_variables if name not in globals()]

if missing:
    raise RuntimeError(
        f"Missing variables: {missing}. "
        "Run your dataset and DataLoader setup cells first."
    )

if not HEAD_CHECKPOINT.exists():
    raise FileNotFoundError(
        f"Checkpoint not found: {HEAD_CHECKPOINT}. "
        "Restore or upload your saved checkpoint first."
    )


# ------------------------------------------------------------
# 3. Load the trained classifier checkpoint
# ------------------------------------------------------------

checkpoint = torch.load(
    HEAD_CHECKPOINT,
    map_location="cpu",
    weights_only=True
)

if checkpoint["num_classes"] != NUM_CLASSES:
    raise ValueError(
        "Checkpoint class count does not match the current dataset."
    )

if checkpoint["class_to_idx"] != {
    str(k): int(v) for k, v in class_to_idx.items()
}:
    raise ValueError(
        "Class mapping mismatch. Use the same class_to_idx mapping "
        "that was used during head training."
    )

model = inception_v3(weights=Inception_V3_Weights.DEFAULT)
model.aux_logits = False
model.AuxLogits = None
model.fc = nn.Linear(model.fc.in_features, NUM_CLASSES)

model.load_state_dict(checkpoint["model_state_dict"])

del checkpoint
gc.collect()

# Freeze every parameter first.
for parameter in model.parameters():
    parameter.requires_grad = False


# ------------------------------------------------------------
# 4. Unfreeze the final Inception blocks and classifier
# ------------------------------------------------------------

for module in [model.Mixed_7a, model.Mixed_7b, model.Mixed_7c, model.fc]:
    for parameter in module.parameters():
        parameter.requires_grad = True

model = model.to(
    device,
    memory_format=torch.channels_last
)

trainable_params = sum(
    p.numel() for p in model.parameters() if p.requires_grad
)
total_params = sum(p.numel() for p in model.parameters())

print(f"Trainable parameters: {trainable_params:,}")
print(f"Total parameters: {total_params:,}")
print(f"Trainable fraction: {100 * trainable_params / total_params:.2f}%")


# ------------------------------------------------------------
# 5. Optimizer and loss
# ------------------------------------------------------------

criterion = nn.CrossEntropyLoss()

optimizer = optim.AdamW(
    [
        {
            "params": [
                p for p in model.parameters()
                if p.requires_grad
                and not any(
                    p is q for q in model.fc.parameters()
                )
            ],
            "lr": BACKBONE_LR,
        },
        {
            "params": model.fc.parameters(),
            "lr": CLASSIFIER_LR,
        },
    ],
    weight_decay=WEIGHT_DECAY,
)

# Reduce learning rate when validation loss stops improving.
scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=2,
    min_lr=1e-7,
)


# ------------------------------------------------------------
# 6. Safe checkpoint saving
# ------------------------------------------------------------

def save_checkpoint_safely(payload, destination):
    destination = Path(destination)
    destination.parent.mkdir(parents=True, exist_ok=True)
    temp_path = destination.with_name(destination.name + ".tmp")

    try:
        safe_payload = dict(payload)

        safe_payload["class_to_idx"] = {
            str(k): int(v)
            for k, v in safe_payload["class_to_idx"].items()
        }
        safe_payload["idx_to_species"] = {
            int(k): str(v)
            for k, v in safe_payload["idx_to_species"].items()
        }
        safe_payload["num_classes"] = int(safe_payload["num_classes"])
        safe_payload["best_val_acc"] = float(safe_payload["best_val_acc"])
        safe_payload["epoch"] = int(safe_payload["epoch"])

        safe_payload["model_state_dict"] = {
            key: tensor.detach().cpu().contiguous().clone()
            for key, tensor in safe_payload["model_state_dict"].items()
        }

        with open(temp_path, "wb") as f:
            torch.save(safe_payload, f)
            f.flush()
            os.fsync(f.fileno())

        verified = torch.load(
            temp_path,
            map_location="cpu",
            weights_only=True
        )

        assert "model_state_dict" in verified
        assert len(verified["model_state_dict"]) > 0
        assert "best_val_acc" in verified
        assert "class_to_idx" in verified

        os.replace(temp_path, destination)

        print(
            f"Verified checkpoint saved: {destination} "
            f"({destination.stat().st_size / 1e6:.1f} MB)"
        )

    except Exception:
        if temp_path.exists():
            temp_path.unlink()
        raise


# ------------------------------------------------------------
# 7. Initialize best-model tracking
# ------------------------------------------------------------

best_val_acc = -1.0
best_val_loss = float("inf")
best_epoch = 0
epochs_without_improvement = 0
PATIENCE = 5

history = []

# Save the starting fine-tuning state as a recovery checkpoint.
initial_payload = {
    "model_state_dict": model.state_dict(),
    "class_to_idx": class_to_idx,
    "idx_to_species": idx_to_species,
    "num_classes": NUM_CLASSES,
    "best_val_acc": -1.0,
    "epoch": 0,
    "architecture": "inception_v3",
    "phase": "final_blocks_finetuning",
}

save_checkpoint_safely(initial_payload, BEST_PATH)
del initial_payload


# ------------------------------------------------------------
# 8. Fine-tuning loop
# ------------------------------------------------------------

for epoch in range(EPOCHS):

    # Keep frozen backbone layers in evaluation mode.
    # Train only the unfrozen final blocks and classifier.
    model.eval()
    model.Mixed_7a.train()
    model.Mixed_7b.train()
    model.Mixed_7c.train()
    model.fc.train()

    train_loss_sum = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in tqdm(
        train_loader,
        desc=f"Epoch {epoch + 1}/{EPOCHS} [Train]"
    ):
        images = images.to(
            device,
            non_blocking=True,
            memory_format=torch.channels_last
        )
        labels = labels.to(device, non_blocking=True).long()

        optimizer.zero_grad(set_to_none=True)

        with torch.autocast("cuda", dtype=torch.bfloat16):
            outputs = model(images)
            loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        batch_size = labels.size(0)
        train_loss_sum += loss.item() * batch_size
        train_correct += (outputs.argmax(1) == labels).sum().item()
        train_total += batch_size

    train_loss = train_loss_sum / train_total
    train_acc = 100.0 * train_correct / train_total

    # Validation
    model.eval()

    val_loss_sum = 0.0
    val_correct = 0
    val_total = 0

    with torch.inference_mode():
        for images, labels in tqdm(
            val_loader,
            desc=f"Epoch {epoch + 1}/{EPOCHS} [Val]",
            leave=False
        ):
            images = images.to(
                device,
                non_blocking=True,
                memory_format=torch.channels_last
            )
            labels = labels.to(device, non_blocking=True).long()

            with torch.autocast("cuda", dtype=torch.bfloat16):
                outputs = model(images)
                loss = criterion(outputs, labels)

            batch_size = labels.size(0)
            val_loss_sum += loss.item() * batch_size
            val_correct += (outputs.argmax(1) == labels).sum().item()
            val_total += batch_size

    val_loss = val_loss_sum / val_total
    val_acc = 100.0 * val_correct / val_total

    scheduler.step(val_loss)

    print(f"\nEpoch {epoch + 1}/{EPOCHS}")
    print(f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%")
    print(f"Val Loss:   {val_loss:.4f} | Val Acc:   {val_acc:.2f}%")
    print(
        "Learning rates: "
        f"backbone={optimizer.param_groups[0]['lr']:.2e}, "
        f"fc={optimizer.param_groups[1]['lr']:.2e}"
    )

    history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_acc": train_acc,
        "val_loss": val_loss,
        "val_acc": val_acc,
    })

    # Select the best model using validation accuracy.
    if val_acc > best_val_acc:
        candidate_state = {
            key: tensor.detach().cpu().clone()
            for key, tensor in model.state_dict().items()
        }

        payload = {
            "model_state_dict": candidate_state,
            "class_to_idx": class_to_idx,
            "idx_to_species": idx_to_species,
            "num_classes": NUM_CLASSES,
            "best_val_acc": float(val_acc),
            "epoch": epoch + 1,
            "architecture": "inception_v3",
            "phase": "final_blocks_finetuning",
        }

        save_checkpoint_safely(payload, BEST_PATH)

        best_val_acc = float(val_acc)
        best_val_loss = float(val_loss)
        best_epoch = epoch + 1
        epochs_without_improvement = 0

        print(f"New best validation accuracy: {best_val_acc:.2f}%")

        del candidate_state, payload
    else:
        epochs_without_improvement += 1

    print("-" * 65)

    if epochs_without_improvement >= PATIENCE:
        print(
            f"Early stopping: validation accuracy has not improved "
            f"for {PATIENCE} consecutive epochs."
        )
        break


# ------------------------------------------------------------
# 9. Restore best checkpoint
# ------------------------------------------------------------

print("\nRestoring best fine-tuned model...")

best_checkpoint = torch.load(
    BEST_PATH,
    map_location="cpu",
    weights_only=True
)

model.load_state_dict(best_checkpoint["model_state_dict"])
model.to(device, memory_format=torch.channels_last)
model.eval()

print("\nFine-tuning complete.")
print(f"Best epoch: {best_checkpoint['epoch']}")
print(f"Best validation accuracy: {best_checkpoint['best_val_acc']:.2f}%")
print(f"Checkpoint: {BEST_PATH}")
print(f"Checkpoint size: {BEST_PATH.stat().st_size / 1e6:.1f} MB")

print(
    "\nDownload this checkpoint before the Modal environment is terminated. "
    "The /root directory is not persistent storage."
)
```

```python

# ============================================================
# Inception V3: Continued Fine-tuning
# Modal | NVIDIA L40S | PyTorch 2.8+
#
# Continues from the best previous fine-tuned checkpoint.
# Lower learning rates, early stopping, macro-F1 evaluation.
# ============================================================

import os
import gc
from pathlib import Path
from collections import defaultdict

import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd

from torchvision.models import inception_v3
from tqdm.auto import tqdm
from sklearn.metrics import f1_score, classification_report


# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

EPOCHS = 10
BACKBONE_LR = 5e-6
CLASSIFIER_LR = 5e-5
WEIGHT_DECAY = 1e-4
PATIENCE = 4

CHECKPOINT_DIR = Path("/root/snake_checkpoints")

SOURCE_PATH = (
    CHECKPOINT_DIR / "inception_v3_snakes_finetuned_best.pth"
)

# Save to a NEW file so the existing best checkpoint is preserved.
BEST_PATH = (
    CHECKPOINT_DIR / "inception_v3_snakes_finetuned_v2_best.pth"
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

if device.type != "cuda":
    raise RuntimeError("CUDA GPU not detected.")

if not torch.cuda.is_bf16_supported():
    raise RuntimeError("This script requires BF16 support.")

print(f"GPU: {torch.cuda.get_device_name(0)}")
print(f"PyTorch: {torch.__version__}")
print(f"BF16 supported: {torch.cuda.is_bf16_supported()}")
print(f"Source checkpoint: {SOURCE_PATH}")
print(f"Output checkpoint: {BEST_PATH}")


# ------------------------------------------------------------
# 2. Validate notebook variables and checkpoint
# ------------------------------------------------------------

required_variables = [
    "train_loader",
    "val_loader",
    "class_to_idx",
    "idx_to_species",
    "NUM_CLASSES",
]

missing = [name for name in required_variables if name not in globals()]

if missing:
    raise RuntimeError(
        f"Missing notebook variables: {missing}. "
        "Run the dataset and DataLoader setup cells first."
    )

if not SOURCE_PATH.exists():
    raise FileNotFoundError(
        f"Checkpoint not found: {SOURCE_PATH}\n"
        "Restore the saved checkpoint to this path before running."
    )

CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

source = torch.load(
    SOURCE_PATH,
    map_location="cpu",
    weights_only=True,
)

expected_mapping = {
    str(k): int(v) for k, v in class_to_idx.items()
}

if int(source["num_classes"]) != int(NUM_CLASSES):
    raise ValueError("Checkpoint class count differs from NUM_CLASSES.")

if source["class_to_idx"] != expected_mapping:
    raise ValueError(
        "Class mapping mismatch. Use the original class_to_idx mapping."
    )

print(f"Starting validation accuracy: {source['best_val_acc']:.2f}%")
print(f"Starting checkpoint epoch: {source['epoch']}")


# ------------------------------------------------------------
# 3. Rebuild architecture and load previous best weights
# ------------------------------------------------------------

# weights=None avoids downloading ImageNet weights again.
# The checkpoint supplies the already-trained weights.
model = inception_v3(weights=None, aux_logits=False)

model.fc = nn.Linear(model.fc.in_features, NUM_CLASSES)

model.load_state_dict(source["model_state_dict"])
del source
gc.collect()

# Freeze all parameters, then unfreeze the final blocks and head.
for parameter in model.parameters():
    parameter.requires_grad = False

for module in (
    model.Mixed_7a,
    model.Mixed_7b,
    model.Mixed_7c,
    model.fc,
):
    for parameter in module.parameters():
        parameter.requires_grad = True

model = model.to(
    device,
    memory_format=torch.channels_last,
)

print(
    "Trainable parameters:",
    f"{sum(p.numel() for p in model.parameters() if p.requires_grad):,}",
)


# ------------------------------------------------------------
# 4. Optimizer and scheduler
# ------------------------------------------------------------

backbone_parameters = [
    p
    for module in (model.Mixed_7a, model.Mixed_7b, model.Mixed_7c)
    for p in module.parameters()
    if p.requires_grad
]

classifier_parameters = [
    p for p in model.fc.parameters() if p.requires_grad
]

optimizer = optim.AdamW(
    [
        {"params": backbone_parameters, "lr": BACKBONE_LR},
        {"params": classifier_parameters, "lr": CLASSIFIER_LR},
    ],
    weight_decay=WEIGHT_DECAY,
)

criterion = nn.CrossEntropyLoss()

scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=2,
    min_lr=1e-7,
)


# ------------------------------------------------------------
# 5. Safe checkpoint saving
# ------------------------------------------------------------

def save_checkpoint_safely(payload, destination):
    destination = Path(destination)
    destination.parent.mkdir(parents=True, exist_ok=True)
    temp_path = destination.with_name(destination.name + ".tmp")

    try:
        safe_payload = dict(payload)

        safe_payload["class_to_idx"] = {
            str(k): int(v)
            for k, v in safe_payload["class_to_idx"].items()
        }
        safe_payload["idx_to_species"] = {
            int(k): str(v)
            for k, v in safe_payload["idx_to_species"].items()
        }
        safe_payload["num_classes"] = int(safe_payload["num_classes"])
        safe_payload["best_val_acc"] = float(safe_payload["best_val_acc"])
        safe_payload["epoch"] = int(safe_payload["epoch"])

        safe_payload["model_state_dict"] = {
            k: v.detach().cpu().contiguous().clone()
            for k, v in safe_payload["model_state_dict"].items()
        }

        with open(temp_path, "wb") as f:
            torch.save(safe_payload, f)
            f.flush()
            os.fsync(f.fileno())

        verified = torch.load(
            temp_path,
            map_location="cpu",
            weights_only=True,
        )

        assert "model_state_dict" in verified
        assert len(verified["model_state_dict"]) > 0
        assert "best_val_acc" in verified
        assert "class_to_idx" in verified

        os.replace(temp_path, destination)

        print(
            f"Verified checkpoint saved: {destination} "
            f"({destination.stat().st_size / 1e6:.1f} MB)"
        )

    except Exception:
        if temp_path.exists():
            temp_path.unlink()
        raise


# ------------------------------------------------------------
# 6. Save starting recovery checkpoint
# ------------------------------------------------------------

# Preserve the starting weights even if no later epoch improves accuracy.
# Use the previous validation score as the initial comparison threshold.
best_val_acc = -1.0
best_epoch = 0
epochs_without_improvement = 0
history_v2 = []

initial_state = {
    k: v.detach().cpu().clone()
    for k, v in model.state_dict().items()
}

initial_payload = {
    "model_state_dict": initial_state,
    "class_to_idx": class_to_idx,
    "idx_to_species": idx_to_species,
    "num_classes": NUM_CLASSES,
    "best_val_acc": -1.0,
    "epoch": 0,
    "architecture": "inception_v3",
    "phase": "continued_finetuning",
}

save_checkpoint_safely(initial_payload, BEST_PATH)

del initial_state, initial_payload
gc.collect()


# ------------------------------------------------------------
# 7. Continued fine-tuning
# ------------------------------------------------------------

for epoch in range(EPOCHS):

    # Frozen backbone stays in eval mode.
    # Train only the final Inception blocks and classification head.
    model.eval()
    model.Mixed_7a.train()
    model.Mixed_7b.train()
    model.Mixed_7c.train()
    model.fc.train()

    train_loss_sum = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in tqdm(
        train_loader,
        desc=f"Epoch {epoch + 1}/{EPOCHS} [Train]",
    ):
        images = images.to(
            device,
            non_blocking=True,
            memory_format=torch.channels_last,
        )
        labels = labels.to(device, non_blocking=True).long()

        optimizer.zero_grad(set_to_none=True)

        with torch.autocast("cuda", dtype=torch.bfloat16):
            outputs = model(images)
            loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        batch_size = labels.size(0)
        train_loss_sum += loss.item() * batch_size
        train_correct += (outputs.argmax(1) == labels).sum().item()
        train_total += batch_size

    train_loss = train_loss_sum / train_total
    train_acc = 100.0 * train_correct / train_total

    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    model.eval()

    val_loss_sum = 0.0
    val_correct = 0
    val_total = 0

    all_labels = []
    all_predictions = []

    with torch.inference_mode():
        for images, labels in tqdm(
            val_loader,
            desc=f"Epoch {epoch + 1}/{EPOCHS} [Val]",
            leave=False,
        ):
            images = images.to(
                device,
                non_blocking=True,
                memory_format=torch.channels_last,
            )
            labels = labels.to(device, non_blocking=True).long()

            with torch.autocast("cuda", dtype=torch.bfloat16):
                outputs = model(images)
                loss = criterion(outputs, labels)

            predictions = outputs.argmax(dim=1)
            batch_size = labels.size(0)

            val_loss_sum += loss.item() * batch_size
            val_correct += (predictions == labels).sum().item()
            val_total += batch_size

            all_labels.extend(labels.cpu().tolist())
            all_predictions.extend(predictions.cpu().tolist())

    val_loss = val_loss_sum / val_total
    val_acc = 100.0 * val_correct / val_total

    macro_f1 = f1_score(
        all_labels,
        all_predictions,
        labels=list(range(NUM_CLASSES)),
        average="macro",
        zero_division=0,
    )

    scheduler.step(val_loss)

    print(f"\nEpoch {epoch + 1}/{EPOCHS}")
    print(f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%")
    print(f"Val Loss:   {val_loss:.4f} | Val Acc:   {val_acc:.2f}%")
    print(f"Macro-F1:   {macro_f1:.4f}")
    print(
        f"LR backbone: {optimizer.param_groups[0]['lr']:.2e} | "
        f"LR head: {optimizer.param_groups[1]['lr']:.2e}"
    )

    history_v2.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_acc": train_acc,
        "val_loss": val_loss,
        "val_acc": val_acc,
        "macro_f1": macro_f1,
    })

    # Save when validation accuracy improves.
    if val_acc > best_val_acc:
        candidate_state = {
            k: v.detach().cpu().clone()
            for k, v in model.state_dict().items()
        }

        payload = {
            "model_state_dict": candidate_state,
            "class_to_idx": class_to_idx,
            "idx_to_species": idx_to_species,
            "num_classes": NUM_CLASSES,
            "best_val_acc": float(val_acc),
            "epoch": epoch + 1,
            "architecture": "inception_v3",
            "phase": "continued_finetuning",
            "val_macro_f1": float(macro_f1),
        }

        save_checkpoint_safely(payload, BEST_PATH)

        best_val_acc = float(val_acc)
        best_epoch = epoch + 1
        epochs_without_improvement = 0

        print(f"New best validation accuracy: {best_val_acc:.2f}%")

        del candidate_state, payload

    else:
        epochs_without_improvement += 1

    print("-" * 65)

    if epochs_without_improvement >= PATIENCE:
        print(
            f"Early stopping after {PATIENCE} epochs without "
            "validation-accuracy improvement."
        )
        break


# ------------------------------------------------------------
# 8. Restore best model and summarize
# ------------------------------------------------------------

best_checkpoint = torch.load(
    BEST_PATH,
    map_location="cpu",
    weights_only=True,
)

model.load_state_dict(best_checkpoint["model_state_dict"])
model.to(device, memory_format=torch.channels_last)
model.eval()

print("\nContinued fine-tuning complete.")
print(f"Best epoch in this run: {best_checkpoint['epoch']}")
print(f"Best validation accuracy: {best_checkpoint['best_val_acc']:.2f}%")
print(f"Checkpoint: {BEST_PATH}")

history_df_v2 = pd.DataFrame(history_v2)
display(history_df_v2)


# ------------------------------------------------------------
# 9. Per-class validation report for the restored best model
# ------------------------------------------------------------

model.eval()
y_true = []
y_pred = []

with torch.inference_mode():
    for images, labels in tqdm(val_loader, desc="Final validation report"):
        images = images.to(
            device,
            non_blocking=True,
            memory_format=torch.channels_last,
        )

        with torch.autocast("cuda", dtype=torch.bfloat16):
            outputs = model(images)

        y_pred.extend(outputs.argmax(1).cpu().tolist())
        y_true.extend(labels.tolist())

report = classification_report(
    y_true,
    y_pred,
    labels=list(range(NUM_CLASSES)),
    output_dict=True,
    zero_division=0,
)

per_class_df_v2 = pd.DataFrame([
    {
        "class_id": int(class_id),
        "species": idx_to_species.get(
            int(class_id),
            idx_to_species.get(str(class_id), "Unknown"),
        ),
        "precision": report[str(class_id)]["precision"],
        "recall": report[str(class_id)]["recall"],
        "f1_score": report[str(class_id)]["f1-score"],
        "support": int(report[str(class_id)]["support"]),
    }
    for class_id in range(NUM_CLASSES)
]).sort_values("f1_score")

print("\n20 lowest-F1 species:")
display(per_class_df_v2.head(20))

per_class_df_v2.to_csv(
    CHECKPOINT_DIR / "per_class_validation_metrics.csv",
    index=False,
)

print(
    "\nDownload the best checkpoint and CSV before the Modal "
    "environment is terminated. Local storage is temporary."
)
```

```python

# ============================================================
# Inception V3: Expanded Fine-tuning for Snake Classification
# Modal | NVIDIA L40S | PyTorch 2.8+
#
# Unfreezes Mixed_6a through Mixed_6e and Mixed_7a through 7c.
# Uses differential learning rates, BF16, early stopping,
# verified checkpoints, validation accuracy and macro-F1.
# ============================================================

import os
import gc
from pathlib import Path

import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd

from torchvision.models import inception_v3
from tqdm.auto import tqdm
from sklearn.metrics import f1_score, classification_report


# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

EPOCHS = 8
PATIENCE = 3
WEIGHT_DECAY = 1e-4

LR_MIXED6 = 1e-6
LR_MIXED7 = 2e-6
LR_FC = 2e-5

CHECKPOINT_DIR = Path("/root/snake_checkpoints")

SOURCE_PATH = (
    CHECKPOINT_DIR / "inception_v3_snakes_finetuned_v2_best.pth"
)

# Never overwrite the previous best checkpoint.
BEST_PATH = (
    CHECKPOINT_DIR / "inception_v3_snakes_finetuned_v3_best.pth"
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

if device.type != "cuda":
    raise RuntimeError("CUDA GPU not detected.")

if not torch.cuda.is_bf16_supported():
    raise RuntimeError("BF16 is not supported by the selected GPU.")

torch.backends.cudnn.benchmark = True

print(f"GPU: {torch.cuda.get_device_name(0)}")
print(f"PyTorch: {torch.__version__}")
print(f"BF16 supported: {torch.cuda.is_bf16_supported()}")
print(f"Source checkpoint: {SOURCE_PATH}")
print(f"New checkpoint: {BEST_PATH}")


# ------------------------------------------------------------
# 2. Validate existing notebook variables
# ------------------------------------------------------------

required_variables = [
    "train_loader",
    "val_loader",
    "class_to_idx",
    "idx_to_species",
    "NUM_CLASSES",
]

missing = [v for v in required_variables if v not in globals()]

if missing:
    raise RuntimeError(
        f"Missing variables: {missing}. "
        "Run your dataset and DataLoader setup cells first."
    )

if not SOURCE_PATH.exists():
    raise FileNotFoundError(
        f"Source checkpoint not found: {SOURCE_PATH}\n"
        "Restore the v2 checkpoint before running this script."
    )

CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

expected_mapping = {
    str(k): int(v) for k, v in class_to_idx.items()
}

# Load the previous best checkpoint.
source = torch.load(
    SOURCE_PATH,
    map_location="cpu",
    weights_only=True,
)

if int(source["num_classes"]) != int(NUM_CLASSES):
    raise ValueError("Checkpoint class count does not match NUM_CLASSES.")

if source["class_to_idx"] != expected_mapping:
    raise ValueError(
        "Class mapping mismatch. Use the original class_to_idx mapping."
    )

baseline_acc = float(source["best_val_acc"])

print(f"Baseline validation accuracy: {baseline_acc:.2f}%")
print(f"Baseline checkpoint epoch: {source['epoch']}")


# ------------------------------------------------------------
# 3. Rebuild architecture and load previous best weights
# ------------------------------------------------------------

# The checkpoint contains the trained weights, so ImageNet
# weights do not need to be downloaded again.
model = inception_v3(weights=None, aux_logits=False)
model.fc = nn.Linear(model.fc.in_features, NUM_CLASSES)

model.load_state_dict(source["model_state_dict"])

del source
gc.collect()

# Freeze the entire network first.
for parameter in model.parameters():
    parameter.requires_grad = False

# Intermediate blocks to unfreeze.
mixed6_modules = [
    model.Mixed_6a,
    model.Mixed_6b,
    model.Mixed_6c,
    model.Mixed_6d,
    model.Mixed_6e,
]

# Final blocks remain trainable.
mixed7_modules = [
    model.Mixed_7a,
    model.Mixed_7b,
    model.Mixed_7c,
]

for module in mixed6_modules + mixed7_modules + [model.fc]:
    for parameter in module.parameters():
        parameter.requires_grad = True

model = model.to(
    device,
    memory_format=torch.channels_last,
)

trainable_params = sum(
    p.numel() for p in model.parameters() if p.requires_grad
)

total_params = sum(p.numel() for p in model.parameters())

print(f"Total parameters: {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
print(f"Trainable fraction: {100 * trainable_params / total_params:.2f}%")


# ------------------------------------------------------------
# 4. Optimizer with differential learning rates
# ------------------------------------------------------------

mixed6_params = [
    p for module in mixed6_modules
    for p in module.parameters()
    if p.requires_grad
]

mixed7_params = [
    p for module in mixed7_modules
    for p in module.parameters()
    if p.requires_grad
]

fc_params = [
    p for p in model.fc.parameters()
    if p.requires_grad
]

optimizer = optim.AdamW(
    [
        {"params": mixed6_params, "lr": LR_MIXED6},
        {"params": mixed7_params, "lr": LR_MIXED7},
        {"params": fc_params, "lr": LR_FC},
    ],
    weight_decay=WEIGHT_DECAY,
)

criterion = nn.CrossEntropyLoss()

scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=2,
    min_lr=1e-7,
)


# ------------------------------------------------------------
# 5. Safe checkpoint saving
# ------------------------------------------------------------

def save_checkpoint_safely(payload, destination):
    destination = Path(destination)
    destination.parent.mkdir(parents=True, exist_ok=True)

    temp_path = destination.with_name(destination.name + ".tmp")

    try:
        safe_payload = dict(payload)

        safe_payload["class_to_idx"] = {
            str(k): int(v)
            for k, v in safe_payload["class_to_idx"].items()
        }

        safe_payload["idx_to_species"] = {
            int(k): str(v)
            for k, v in safe_payload["idx_to_species"].items()
        }

        safe_payload["num_classes"] = int(safe_payload["num_classes"])
        safe_payload["best_val_acc"] = float(safe_payload["best_val_acc"])
        safe_payload["epoch"] = int(safe_payload["epoch"])

        safe_payload["model_state_dict"] = {
            k: v.detach().cpu().contiguous().clone()
            for k, v in safe_payload["model_state_dict"].items()
        }

        with open(temp_path, "wb") as f:
            torch.save(safe_payload, f)
            f.flush()
            os.fsync(f.fileno())

        # Verify the complete file using restricted loading.
        verified = torch.load(
            temp_path,
            map_location="cpu",
            weights_only=True,
        )

        assert "model_state_dict" in verified
        assert len(verified["model_state_dict"]) > 0
        assert "best_val_acc" in verified
        assert "class_to_idx" in verified

        os.replace(temp_path, destination)

        print(
            f"Verified checkpoint saved: {destination} "
            f"({destination.stat().st_size / 1e6:.1f} MB)"
        )

    except Exception:
        if temp_path.exists():
            temp_path.unlink()
        raise


# ------------------------------------------------------------
# 6. Preserve baseline as the initial candidate
# ------------------------------------------------------------

# Copy the original checkpoint to the new path. If no epoch
# beats the baseline, the original best weights remain available
# under the new filename.
baseline_checkpoint = torch.load(
    SOURCE_PATH,
    map_location="cpu",
    weights_only=True,
)

initial_payload = {
    "model_state_dict": baseline_checkpoint["model_state_dict"],
    "class_to_idx": class_to_idx,
    "idx_to_species": idx_to_species,
    "num_classes": NUM_CLASSES,
    "best_val_acc": baseline_acc,
    "epoch": int(baseline_checkpoint["epoch"]),
    "architecture": "inception_v3",
    "phase": "expanded_finetuning",
    "baseline_val_acc": baseline_acc,
}

save_checkpoint_safely(initial_payload, BEST_PATH)

del initial_payload, baseline_checkpoint
gc.collect()

best_val_acc = baseline_acc
best_epoch = 0
epochs_without_improvement = 0
history_v3 = []


# ------------------------------------------------------------
# 7. Fine-tuning loop
# ------------------------------------------------------------

for epoch in range(EPOCHS):

    # Train unfrozen layers but keep BatchNorm in evaluation mode.
    # This prevents running-statistic drift while fine-tuning.
    model.eval()

    for module in mixed6_modules + mixed7_modules + [model.fc]:
        module.train()

    for module in model.modules():
        if isinstance(module, nn.modules.batchnorm._BatchNorm):
            module.eval()

    train_loss_sum = 0.0
    train_correct = 0
    train_total = 0

    train_bar = tqdm(
        train_loader,
        desc=f"Epoch {epoch + 1}/{EPOCHS} [Train]",
    )

    for images, labels in train_bar:
        images = images.to(
            device,
            non_blocking=True,
            memory_format=torch.channels_last,
        )
        labels = labels.to(device, non_blocking=True).long()

        optimizer.zero_grad(set_to_none=True)

        with torch.autocast("cuda", dtype=torch.bfloat16):
            outputs = model(images)
            loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        batch_size = labels.size(0)
        train_loss_sum += loss.item() * batch_size
        train_correct += (outputs.argmax(1) == labels).sum().item()
        train_total += batch_size

        train_bar.set_postfix(
            loss=f"{loss.item():.4f}",
            acc=f"{100 * train_correct / train_total:.2f}%",
        )

    train_loss = train_loss_sum / train_total
    train_acc = 100.0 * train_correct / train_total

    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    model.eval()

    val_loss_sum = 0.0
    val_correct = 0
    val_total = 0

    all_labels = []
    all_predictions = []

    with torch.inference_mode():
        for images, labels in tqdm(
            val_loader,
            desc=f"Epoch {epoch + 1}/{EPOCHS} [Val]",
            leave=False,
        ):
            images = images.to(
                device,
                non_blocking=True,
                memory_format=torch.channels_last,
            )
            labels = labels.to(device, non_blocking=True).long()

            with torch.autocast("cuda", dtype=torch.bfloat16):
                outputs = model(images)
                loss = criterion(outputs, labels)

            predictions = outputs.argmax(dim=1)
            batch_size = labels.size(0)

            val_loss_sum += loss.item() * batch_size
            val_correct += (predictions == labels).sum().item()
            val_total += batch_size

            all_labels.extend(labels.cpu().tolist())
            all_predictions.extend(predictions.cpu().tolist())

    val_loss = val_loss_sum / val_total
    val_acc = 100.0 * val_correct / val_total

    macro_f1 = f1_score(
        all_labels,
        all_predictions,
        labels=list(range(NUM_CLASSES)),
        average="macro",
        zero_division=0,
    )

    scheduler.step(val_loss)

    print(f"\nEpoch {epoch + 1}/{EPOCHS}")
    print(f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%")
    print(f"Val Loss:   {val_loss:.4f} | Val Acc:   {val_acc:.2f}%")
    print(f"Macro-F1:   {macro_f1:.4f}")
    print(
        f"LR Mixed6: {optimizer.param_groups[0]['lr']:.2e} | "
        f"LR Mixed7: {optimizer.param_groups[1]['lr']:.2e} | "
        f"LR FC: {optimizer.param_groups[2]['lr']:.2e}"
    )

    history_v3.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_acc": train_acc,
        "val_loss": val_loss,
        "val_acc": val_acc,
        "macro_f1": macro_f1,
    })

    # Promote the model only if it beats the previous best.
    if val_acc > best_val_acc:

        candidate_state = {
            k: v.detach().cpu().clone()
            for k, v in model.state_dict().items()
        }

        payload = {
            "model_state_dict": candidate_state,
            "class_to_idx": class_to_idx,
            "idx_to_species": idx_to_species,
            "num_classes": NUM_CLASSES,
            "best_val_acc": float(val_acc),
            "epoch": epoch + 1,
            "architecture": "inception_v3",
            "phase": "expanded_finetuning",
            "baseline_val_acc": baseline_acc,
            "val_macro_f1": float(macro_f1),
        }

        save_checkpoint_safely(payload, BEST_PATH)

        best_val_acc = float(val_acc)
        best_epoch = epoch + 1
        epochs_without_improvement = 0

        print(
            f"New overall best validation accuracy: "
            f"{best_val_acc:.2f}%"
        )

        del candidate_state, payload

    else:
        epochs_without_improvement += 1
        print(
            f"No new accuracy record. "
            f"Best remains {best_val_acc:.2f}%."
        )

    print("-" * 70)

    if epochs_without_improvement >= PATIENCE:
        print(
            f"Early stopping: no new best validation accuracy "
            f"for {PATIENCE} consecutive epochs."
        )
        break


# ------------------------------------------------------------
# 8. Restore the best model, including the old baseline if needed
# ------------------------------------------------------------

best_checkpoint = torch.load(
    BEST_PATH,
    map_location="cpu",
    weights_only=True,
)

model.load_state_dict(best_checkpoint["model_state_dict"])
model.to(device, memory_format=torch.channels_last)
model.eval()

print("\nExpanded fine-tuning complete.")
print(f"Baseline accuracy: {baseline_acc:.2f}%")
print(f"Best accuracy retained: {best_checkpoint['best_val_acc']:.2f}%")
print(f"Best epoch in this run: {best_checkpoint['epoch']}")
print(f"Checkpoint: {BEST_PATH}")

history_df_v3 = pd.DataFrame(history_v3)
display(history_df_v3)


# ------------------------------------------------------------
# 9. Per-class validation metrics for the restored best model
# ------------------------------------------------------------

model.eval()

y_true = []
y_pred = []

with torch.inference_mode():
    for images, labels in tqdm(
        val_loader,
        desc="Final validation report",
    ):
        images = images.to(
            device,
            non_blocking=True,
            memory_format=torch.channels_last,
        )

        with torch.autocast("cuda", dtype=torch.bfloat16):
            outputs = model(images)

        y_pred.extend(outputs.argmax(1).cpu().tolist())
        y_true.extend(labels.tolist())

report = classification_report(
    y_true,
    y_pred,
    labels=list(range(NUM_CLASSES)),
    output_dict=True,
    zero_division=0,
)

per_class_df_v3 = pd.DataFrame([
    {
        "class_id": class_id,
        "species": idx_to_species.get(
            class_id,
            idx_to_species.get(str(class_id), "Unknown"),
        ),
        "precision": report[str(class_id)]["precision"],
        "recall": report[str(class_id)]["recall"],
        "f1_score": report[str(class_id)]["f1-score"],
        "support": int(report[str(class_id)]["support"]),
    }
    for class_id in range(NUM_CLASSES)
]).sort_values("f1_score")

print("\n20 lowest-F1 species:")
display(per_class_df_v3.head(20))

CSV_PATH = CHECKPOINT_DIR / "per_class_validation_metrics_v3.csv"
per_class_df_v3.to_csv(CSV_PATH, index=False)

print(f"\nPer-class report: {CSV_PATH}")
print(
    "\nDownload the checkpoint and CSV before the Modal environment "
    "is terminated. The /root directory is ephemeral."
)
```

```python

# ============================================================
# Inception V3: Fine-tuning V4
# Modal | NVIDIA L40S | BF16
#
# Resume from v3 best checkpoint (50.29% validation accuracy)
# Keep Mixed_6a-6e, Mixed_7a-7c and fc trainable.
# Lower learning rates, early stopping, verified checkpoints.
# ============================================================

import os
import gc
from pathlib import Path

import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd

from torchvision.models import inception_v3
from tqdm.auto import tqdm
from sklearn.metrics import f1_score, classification_report


# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

EPOCHS = 8
PATIENCE = 3
WEIGHT_DECAY = 1e-4

LR_MIXED6 = 5e-7
LR_MIXED7 = 1e-6
LR_FC = 1e-5

CHECKPOINT_DIR = Path("/root/snake_checkpoints")

SOURCE_PATH = (
    CHECKPOINT_DIR / "inception_v3_snakes_finetuned_v3_best.pth"
)

# Do not overwrite the current best checkpoint.
BEST_PATH = (
    CHECKPOINT_DIR / "inception_v3_snakes_finetuned_v4_best.pth"
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

if device.type != "cuda":
    raise RuntimeError("CUDA GPU not detected.")

if not torch.cuda.is_bf16_supported():
    raise RuntimeError("BF16 is not supported by this GPU.")

torch.backends.cudnn.benchmark = True

print(f"GPU: {torch.cuda.get_device_name(0)}")
print(f"PyTorch: {torch.__version__}")
print(f"Source: {SOURCE_PATH}")
print(f"Output: {BEST_PATH}")


# ------------------------------------------------------------
# 2. Validate environment and source checkpoint
# ------------------------------------------------------------

required_variables = [
    "train_loader",
    "val_loader",
    "class_to_idx",
    "idx_to_species",
    "NUM_CLASSES",
]

missing = [name for name in required_variables if name not in globals()]

if missing:
    raise RuntimeError(
        f"Missing variables: {missing}. "
        "Run your dataset and DataLoader setup first."
    )

if not SOURCE_PATH.exists():
    raise FileNotFoundError(
        f"Checkpoint not found: {SOURCE_PATH}. "
        "Restore your v3 checkpoint before continuing."
    )

CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

source = torch.load(
    SOURCE_PATH,
    map_location="cpu",
    weights_only=True,
)

expected_mapping = {
    str(k): int(v) for k, v in class_to_idx.items()
}

if int(source["num_classes"]) != int(NUM_CLASSES):
    raise ValueError("Checkpoint class count does not match NUM_CLASSES.")

if source["class_to_idx"] != expected_mapping:
    raise ValueError(
        "Class mapping mismatch. Use the original class_to_idx mapping."
    )

baseline_acc = float(source["best_val_acc"])

print(f"Baseline validation accuracy: {baseline_acc:.2f}%")
print(f"Source checkpoint epoch: {source['epoch']}")


# ------------------------------------------------------------
# 3. Rebuild model and restore v3 weights
# ------------------------------------------------------------

model = inception_v3(weights=None, aux_logits=False)
model.fc = nn.Linear(model.fc.in_features, NUM_CLASSES)

model.load_state_dict(source["model_state_dict"])

del source
gc.collect()

# Freeze all layers first.
for parameter in model.parameters():
    parameter.requires_grad = False

mixed6_modules = [
    model.Mixed_6a,
    model.Mixed_6b,
    model.Mixed_6c,
    model.Mixed_6d,
    model.Mixed_6e,
]

mixed7_modules = [
    model.Mixed_7a,
    model.Mixed_7b,
    model.Mixed_7c,
]

for module in mixed6_modules + mixed7_modules + [model.fc]:
    for parameter in module.parameters():
        parameter.requires_grad = True

model = model.to(
    device,
    memory_format=torch.channels_last,
)

print(
    "Trainable parameters:",
    f"{sum(p.numel() for p in model.parameters() if p.requires_grad):,}",
)


# ------------------------------------------------------------
# 4. Optimizer and scheduler
# ------------------------------------------------------------

optimizer = optim.AdamW(
    [
        {
            "params": [
                p for module in mixed6_modules
                for p in module.parameters()
                if p.requires_grad
            ],
            "lr": LR_MIXED6,
        },
        {
            "params": [
                p for module in mixed7_modules
                for p in module.parameters()
                if p.requires_grad
            ],
            "lr": LR_MIXED7,
        },
        {
            "params": list(model.fc.parameters()),
            "lr": LR_FC,
        },
    ],
    weight_decay=WEIGHT_DECAY,
)

criterion = nn.CrossEntropyLoss()

scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=2,
    min_lr=1e-8,
)


# ------------------------------------------------------------
# 5. Verified checkpoint saving
# ------------------------------------------------------------

def save_checkpoint_safely(payload, destination):
    destination = Path(destination)
    destination.parent.mkdir(parents=True, exist_ok=True)

    temp_path = destination.with_name(destination.name + ".tmp")

    try:
        safe_payload = dict(payload)

        safe_payload["class_to_idx"] = {
            str(k): int(v)
            for k, v in safe_payload["class_to_idx"].items()
        }
        safe_payload["idx_to_species"] = {
            int(k): str(v)
            for k, v in safe_payload["idx_to_species"].items()
        }
        safe_payload["num_classes"] = int(safe_payload["num_classes"])
        safe_payload["best_val_acc"] = float(safe_payload["best_val_acc"])
        safe_payload["epoch"] = int(safe_payload["epoch"])

        safe_payload["model_state_dict"] = {
            k: v.detach().cpu().contiguous().clone()
            for k, v in safe_payload["model_state_dict"].items()
        }

        with open(temp_path, "wb") as f:
            torch.save(safe_payload, f)
            f.flush()
            os.fsync(f.fileno())

        verified = torch.load(
            temp_path,
            map_location="cpu",
            weights_only=True,
        )

        assert "model_state_dict" in verified
        assert len(verified["model_state_dict"]) > 0
        assert "best_val_acc" in verified
        assert "class_to_idx" in verified

        os.replace(temp_path, destination)

        print(
            f"Verified checkpoint saved: {destination} "
            f"({destination.stat().st_size / 1e6:.1f} MB)"
        )

    except Exception:
        if temp_path.exists():
            temp_path.unlink()
        raise


# ------------------------------------------------------------
# 6. Preserve v3 as the starting baseline
# ------------------------------------------------------------

initial_payload = {
    "model_state_dict": model.state_dict(),
    "class_to_idx": class_to_idx,
    "idx_to_species": idx_to_species,
    "num_classes": NUM_CLASSES,
    "best_val_acc": baseline_acc,
    "epoch": int(torch.load(
        SOURCE_PATH, map_location="cpu", weights_only=True
    )["epoch"]),
    "architecture": "inception_v3",
    "phase": "continued_finetuning_v4",
    "baseline_val_acc": baseline_acc,
}

save_checkpoint_safely(initial_payload, BEST_PATH)

del initial_payload
gc.collect()

best_val_acc = baseline_acc
best_epoch = 0
epochs_without_improvement = 0
history_v4 = []


# ------------------------------------------------------------
# 7. Training loop
# ------------------------------------------------------------

for epoch in range(EPOCHS):

    model.eval()

    # Enable training only for unfrozen blocks and the classifier.
    for module in mixed6_modules + mixed7_modules + [model.fc]:
        module.train()

    # Keep BatchNorm running statistics fixed.
    for module in model.modules():
        if isinstance(module, nn.modules.batchnorm._BatchNorm):
            module.eval()

    train_loss_sum = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in tqdm(
        train_loader,
        desc=f"Epoch {epoch + 1}/{EPOCHS} [Train]",
    ):
        images = images.to(
            device,
            non_blocking=True,
            memory_format=torch.channels_last,
        )
        labels = labels.to(device, non_blocking=True).long()

        optimizer.zero_grad(set_to_none=True)

        with torch.autocast("cuda", dtype=torch.bfloat16):
            outputs = model(images)
            loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        batch_size = labels.size(0)
        train_loss_sum += loss.item() * batch_size
        train_correct += (outputs.argmax(1) == labels).sum().item()
        train_total += batch_size

    train_loss = train_loss_sum / train_total
    train_acc = 100.0 * train_correct / train_total

    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    model.eval()

    val_loss_sum = 0.0
    val_correct = 0
    val_total = 0

    y_true = []
    y_pred = []

    with torch.inference_mode():
        for images, labels in tqdm(
            val_loader,
            desc=f"Epoch {epoch + 1}/{EPOCHS} [Val]",
            leave=False,
        ):
            images = images.to(
                device,
                non_blocking=True,
                memory_format=torch.channels_last,
            )
            labels = labels.to(device, non_blocking=True).long()

            with torch.autocast("cuda", dtype=torch.bfloat16):
                outputs = model(images)
                loss = criterion(outputs, labels)

            predictions = outputs.argmax(dim=1)
            batch_size = labels.size(0)

            val_loss_sum += loss.item() * batch_size
            val_correct += (predictions == labels).sum().item()
            val_total += batch_size

            y_true.extend(labels.cpu().tolist())
            y_pred.extend(predictions.cpu().tolist())

    val_loss = val_loss_sum / val_total
    val_acc = 100.0 * val_correct / val_total

    macro_f1 = f1_score(
        y_true,
        y_pred,
        labels=list(range(NUM_CLASSES)),
        average="macro",
        zero_division=0,
    )

    scheduler.step(val_loss)

    print(f"\nEpoch {epoch + 1}/{EPOCHS}")
    print(f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%")
    print(f"Val Loss:   {val_loss:.4f} | Val Acc:   {val_acc:.2f}%")
    print(f"Macro-F1:   {macro_f1:.4f}")
    print(
        f"LR Mixed6: {optimizer.param_groups[0]['lr']:.2e} | "
        f"LR Mixed7: {optimizer.param_groups[1]['lr']:.2e} | "
        f"LR FC: {optimizer.param_groups[2]['lr']:.2e}"
    )

    history_v4.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_acc": train_acc,
        "val_loss": val_loss,
        "val_acc": val_acc,
        "macro_f1": macro_f1,
    })

    # Only promote a checkpoint if it beats the 50.29% baseline.
    if val_acc > best_val_acc:
        candidate_state = {
            k: v.detach().cpu().clone()
            for k, v in model.state_dict().items()
        }

        payload = {
            "model_state_dict": candidate_state,
            "class_to_idx": class_to_idx,
            "idx_to_species": idx_to_species,
            "num_classes": NUM_CLASSES,
            "best_val_acc": float(val_acc),
            "epoch": epoch + 1,
            "architecture": "inception_v3",
            "phase": "continued_finetuning_v4",
            "baseline_val_acc": baseline_acc,
            "val_macro_f1": float(macro_f1),
        }

        save_checkpoint_safely(payload, BEST_PATH)

        best_val_acc = float(val_acc)
        best_epoch = epoch + 1
        epochs_without_improvement = 0

        print(f"New overall best validation accuracy: {best_val_acc:.2f}%")

        del candidate_state, payload
    else:
        epochs_without_improvement += 1
        print(f"No improvement. Best remains {best_val_acc:.2f}%.")

    print("-" * 70)

    if epochs_without_improvement >= PATIENCE:
        print("Early stopping triggered.")
        break


# ------------------------------------------------------------
# 8. Restore best model
# ------------------------------------------------------------

best_checkpoint = torch.load(
    BEST_PATH,
    map_location="cpu",
    weights_only=True,
)

model.load_state_dict(best_checkpoint["model_state_dict"])
model.to(device, memory_format=torch.channels_last)
model.eval()

print("\nFine-tuning V4 complete.")
print(f"Baseline accuracy: {baseline_acc:.2f}%")
print(f"Best accuracy retained: {best_checkpoint['best_val_acc']:.2f}%")
print(f"Best epoch in this run: {best_checkpoint['epoch']}")
print(f"Checkpoint: {BEST_PATH}")

history_df_v4 = pd.DataFrame(history_v4)
display(history_df_v4)


# ------------------------------------------------------------
# 9. Per-class validation report
# ------------------------------------------------------------

y_true = []
y_pred = []

with torch.inference_mode():
    for images, labels in tqdm(val_loader, desc="Final validation report"):
        images = images.to(
            device,
            non_blocking=True,
            memory_format=torch.channels_last,
        )

        with torch.autocast("cuda", dtype=torch.bfloat16):
            outputs = model(images)

        y_pred.extend(outputs.argmax(1).cpu().tolist())
        y_true.extend(labels.tolist())

report = classification_report(
    y_true,
    y_pred,
    labels=list(range(NUM_CLASSES)),
    output_dict=True,
    zero_division=0,
)

per_class_df_v4 = pd.DataFrame([
    {
        "class_id": class_id,
        "species": idx_to_species.get(
            class_id,
            idx_to_species.get(str(class_id), "Unknown"),
        ),
        "precision": report[str(class_id)]["precision"],
        "recall": report[str(class_id)]["recall"],
        "f1_score": report[str(class_id)]["f1-score"],
        "support": int(report[str(class_id)]["support"]),
    }
    for class_id in range(NUM_CLASSES)
]).sort_values("f1_score")

print("\n20 lowest-F1 species:")
display(per_class_df_v4.head(20))

CSV_PATH = CHECKPOINT_DIR / "per_class_validation_metrics_v4.csv"
per_class_df_v4.to_csv(CSV_PATH, index=False)

print(f"Per-class report: {CSV_PATH}")
print(
    "\nDownload your best checkpoint and CSV before the Modal "
    "environment is terminated. Local storage is temporary."
)
```

```python

# ============================================================
# ConvNeXt-Tiny: Snake Species Classification
# Modal | NVIDIA L40S | PyTorch 2.8+
#
# Dataset and existing DataLoaders remain unchanged.
#
# Phase 1: Train classifier head.
# Phase 2: Fine-tune the final ConvNeXt stages.
#
# Features:
# - ImageNet pretrained weights
# - BF16 mixed precision
# - Channels-last memory format
# - Differential learning rates
# - Validation-based checkpoint selection
# - Early stopping and macro-F1 tracking
# - Verified, atomic checkpoint saves
# ============================================================

import os
import gc
from pathlib import Path

import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd

from torchvision.models import convnext_tiny, ConvNeXt_Tiny_Weights
from tqdm.auto import tqdm
from sklearn.metrics import f1_score, classification_report


# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

HEAD_EPOCHS = 3
FINETUNE_EPOCHS = 10
PATIENCE = 4

WEIGHT_DECAY = 1e-4

HEAD_LR = 1e-3
STAGE3_LR = 1e-5
STAGE4_LR = 2e-5
CLASSIFIER_LR = 1e-4

CHECKPOINT_DIR = Path("/root/snake_checkpoints")
CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

BEST_PATH = CHECKPOINT_DIR / "convnext_tiny_snakes_best.pth"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

if device.type != "cuda":
    raise RuntimeError("CUDA GPU not detected.")

if not torch.cuda.is_bf16_supported():
    raise RuntimeError("BF16 is not supported by this GPU.")

torch.backends.cudnn.benchmark = True

print(f"GPU: {torch.cuda.get_device_name(0)}")
print(f"PyTorch: {torch.__version__}")
print(f"BF16 supported: {torch.cuda.is_bf16_supported()}")
print(f"Checkpoint destination: {BEST_PATH}")


# ------------------------------------------------------------
# 2. Validate existing dataset variables
# ------------------------------------------------------------

required_variables = [
    "train_loader",
    "val_loader",
    "class_to_idx",
    "idx_to_species",
    "NUM_CLASSES",
]

missing = [
    name for name in required_variables
    if name not in globals()
]

if missing:
    raise RuntimeError(
        f"Missing variables: {missing}. "
        "Run your existing dataset and DataLoader setup cells first."
    )

if len(class_to_idx) != NUM_CLASSES:
    raise ValueError(
        f"Expected {NUM_CLASSES} classes, "
        f"but class_to_idx contains {len(class_to_idx)}."
    )

class_mapping = {
    str(k): int(v)
    for k, v in class_to_idx.items()
}

species_mapping = {
    int(k): str(v)
    for k, v in idx_to_species.items()
}

print(f"Number of classes: {NUM_CLASSES}")
print(f"Training batches: {len(train_loader)}")
print(f"Validation batches: {len(val_loader)}")


# ------------------------------------------------------------
# 3. Initialize pretrained ConvNeXt-Tiny
# ------------------------------------------------------------

print("\nLoading ImageNet-pretrained ConvNeXt-Tiny...")

weights = ConvNeXt_Tiny_Weights.DEFAULT
model = convnext_tiny(weights=weights)

# ConvNeXt-Tiny classifier:
# classifier[0] = LayerNorm2d
# classifier[1] = Flatten
# classifier[2] = Linear classification layer

in_features = model.classifier[2].in_features

model.classifier[2] = nn.Linear(
    in_features,
    NUM_CLASSES
)

model = model.to(
    device,
    memory_format=torch.channels_last
)

criterion = nn.CrossEntropyLoss()


# ------------------------------------------------------------
# 4. Verified checkpoint saving
# ------------------------------------------------------------

def save_checkpoint_safely(payload, destination):
    destination = Path(destination)
    destination.parent.mkdir(parents=True, exist_ok=True)

    temp_path = destination.with_name(
        destination.name + ".tmp"
    )

    try:
        safe_payload = dict(payload)

        # Convert metadata to standard Python types.
        safe_payload["class_to_idx"] = {
            str(k): int(v)
            for k, v in safe_payload["class_to_idx"].items()
        }

        safe_payload["idx_to_species"] = {
            int(k): str(v)
            for k, v in safe_payload["idx_to_species"].items()
        }

        safe_payload["num_classes"] = int(
            safe_payload["num_classes"]
        )

        safe_payload["best_val_acc"] = float(
            safe_payload["best_val_acc"]
        )

        safe_payload["epoch"] = int(
            safe_payload["epoch"]
        )

        safe_payload["model_state_dict"] = {
            k: v.detach().cpu().contiguous().clone()
            for k, v in safe_payload["model_state_dict"].items()
        }

        with open(temp_path, "wb") as f:
            torch.save(safe_payload, f)
            f.flush()
            os.fsync(f.fileno())

        # Verify the saved checkpoint.
        verified = torch.load(
            temp_path,
            map_location="cpu",
            weights_only=True
        )

        assert "model_state_dict" in verified
        assert len(verified["model_state_dict"]) > 0
        assert "best_val_acc" in verified
        assert "class_to_idx" in verified

        # Replace only after successful verification.
        os.replace(temp_path, destination)

        print(
            f"Verified checkpoint saved: {destination} "
            f"({destination.stat().st_size / 1e6:.1f} MB)"
        )

    except Exception:
        if temp_path.exists():
            temp_path.unlink()
        raise


# ------------------------------------------------------------
# 5. Validation function
# ------------------------------------------------------------

def validate_model(model):
    model.eval()

    total_loss = 0.0
    total_correct = 0
    total_samples = 0

    y_true = []
    y_pred = []

    with torch.inference_mode():
        for images, labels in tqdm(
            val_loader,
            desc="Validation",
            leave=False
        ):
            images = images.to(
                device,
                non_blocking=True,
                memory_format=torch.channels_last
            )

            labels = labels.to(
                device,
                non_blocking=True
            ).long()

            with torch.autocast(
                device_type="cuda",
                dtype=torch.bfloat16
            ):
                outputs = model(images)
                loss = criterion(outputs, labels)

            predictions = outputs.argmax(dim=1)

            batch_size = labels.size(0)

            total_loss += loss.item() * batch_size
            total_correct += (
                predictions == labels
            ).sum().item()

            total_samples += batch_size

            y_true.extend(labels.cpu().tolist())
            y_pred.extend(predictions.cpu().tolist())

    val_loss = total_loss / total_samples
    val_acc = 100.0 * total_correct / total_samples

    macro_f1 = f1_score(
        y_true,
        y_pred,
        labels=list(range(NUM_CLASSES)),
        average="macro",
        zero_division=0
    )

    return val_loss, val_acc, macro_f1, y_true, y_pred


# ------------------------------------------------------------
# 6. Shared training function
# ------------------------------------------------------------

best_val_acc = -1.0
best_epoch = 0
epochs_without_improvement = 0
global_epoch = 0

history_convnext = []


def train_epoch(model, optimizer, epoch_label):
    model.train()

    train_loss_sum = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in tqdm(
        train_loader,
        desc=epoch_label
    ):
        images = images.to(
            device,
            non_blocking=True,
            memory_format=torch.channels_last
        )

        labels = labels.to(
            device,
            non_blocking=True
        ).long()

        optimizer.zero_grad(set_to_none=True)

        with torch.autocast(
            device_type="cuda",
            dtype=torch.bfloat16
        ):
            outputs = model(images)
            loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        batch_size = labels.size(0)

        train_loss_sum += loss.item() * batch_size
        train_correct += (
            outputs.argmax(dim=1) == labels
        ).sum().item()

        train_total += batch_size

    return (
        train_loss_sum / train_total,
        100.0 * train_correct / train_total
    )


def save_if_best(
    model,
    val_acc,
    val_loss,
    macro_f1,
    epoch,
    phase
):
    global best_val_acc, best_epoch
    global epochs_without_improvement

    if val_acc > best_val_acc:
        payload = {
            "model_state_dict": model.state_dict(),
            "class_to_idx": class_mapping,
            "idx_to_species": species_mapping,
            "num_classes": NUM_CLASSES,
            "best_val_acc": float(val_acc),
            "val_loss": float(val_loss),
            "val_macro_f1": float(macro_f1),
            "epoch": int(epoch),
            "architecture": "convnext_tiny",
            "phase": phase,
        }

        save_checkpoint_safely(payload, BEST_PATH)

        best_val_acc = float(val_acc)
        best_epoch = int(epoch)
        epochs_without_improvement = 0

        print(
            f"New best validation accuracy: "
            f"{best_val_acc:.2f}% | "
            f"Macro-F1: {macro_f1:.4f}"
        )

        del payload
        return True

    epochs_without_improvement += 1

    print(
        f"No new best. Current best: "
        f"{best_val_acc:.2f}%"
    )

    return False


# ------------------------------------------------------------
# 7. Phase 1: Train classification head
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("PHASE 1: CLASSIFIER-HEAD TRAINING")
print("=" * 65)

for parameter in model.features.parameters():
    parameter.requires_grad = False

for parameter in model.classifier.parameters():
    parameter.requires_grad = True

optimizer = optim.AdamW(
    model.classifier.parameters(),
    lr=HEAD_LR,
    weight_decay=WEIGHT_DECAY
)

for epoch in range(HEAD_EPOCHS):
    global_epoch += 1

    train_loss, train_acc = train_epoch(
        model,
        optimizer,
        f"Head epoch {epoch + 1}/{HEAD_EPOCHS}"
    )

    val_loss, val_acc, macro_f1, _, _ = validate_model(model)

    print(f"\nHead epoch {epoch + 1}/{HEAD_EPOCHS}")
    print(f"Train loss: {train_loss:.4f}")
    print(f"Train accuracy: {train_acc:.2f}%")
    print(f"Validation loss: {val_loss:.4f}")
    print(f"Validation accuracy: {val_acc:.2f}%")
    print(f"Macro-F1: {macro_f1:.4f}")

    history_convnext.append({
        "epoch": global_epoch,
        "phase": "head",
        "train_loss": train_loss,
        "train_acc": train_acc,
        "val_loss": val_loss,
        "val_acc": val_acc,
        "macro_f1": macro_f1
    })

    save_if_best(
        model,
        val_acc,
        val_loss,
        macro_f1,
        global_epoch,
        "head"
    )


# ------------------------------------------------------------
# 8. Phase 2: Fine-tune the final feature stages
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("PHASE 2: FINAL-STAGE FINE-TUNING")
print("=" * 65)

# ConvNeXt-Tiny features are structured as:
# features[0]: stem
# features[1]: stage 1
# features[2]: downsampling
# features[3]: stage 2
# features[4]: downsampling
# features[5]: stage 3
# features[6]: downsampling
# features[7]: stage 4

# Freeze the backbone again.
for parameter in model.features.parameters():
    parameter.requires_grad = False

# Unfreeze the last two feature stages and their connecting
# downsampling layer.
for index in [5, 6, 7]:
    for parameter in model.features[index].parameters():
        parameter.requires_grad = True

for parameter in model.classifier.parameters():
    parameter.requires_grad = True

optimizer = optim.AdamW(
    [
        {
            "params": model.features[5].parameters(),
            "lr": STAGE3_LR
        },
        {
            "params": model.features[6].parameters(),
            "lr": STAGE3_LR
        },
        {
            "params": model.features[7].parameters(),
            "lr": STAGE4_LR
        },
        {
            "params": model.classifier.parameters(),
            "lr": CLASSIFIER_LR
        }
    ],
    weight_decay=WEIGHT_DECAY
)

scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=2,
    min_lr=1e-7
)

epochs_without_improvement = 0

for epoch in range(FINETUNE_EPOCHS):
    global_epoch += 1

    train_loss, train_acc = train_epoch(
        model,
        optimizer,
        f"Fine-tune {epoch + 1}/{FINETUNE_EPOCHS}"
    )

    val_loss, val_acc, macro_f1, _, _ = validate_model(model)

    scheduler.step(val_loss)

    print(f"\nFine-tuning epoch {epoch + 1}/{FINETUNE_EPOCHS}")
    print(f"Train loss: {train_loss:.4f}")
    print(f"Train accuracy: {train_acc:.2f}%")
    print(f"Validation loss: {val_loss:.4f}")
    print(f"Validation accuracy: {val_acc:.2f}%")
    print(f"Macro-F1: {macro_f1:.4f}")

    print(
        "Learning rates: "
        f"stage3={optimizer.param_groups[0]['lr']:.2e}, "
        f"downsample={optimizer.param_groups[1]['lr']:.2e}, "
        f"stage4={optimizer.param_groups[2]['lr']:.2e}, "
        f"classifier={optimizer.param_groups[3]['lr']:.2e}"
    )

    history_convnext.append({
        "epoch": global_epoch,
        "phase": "fine_tune",
        "train_loss": train_loss,
        "train_acc": train_acc,
        "val_loss": val_loss,
        "val_acc": val_acc,
        "macro_f1": macro_f1
    })

    improved = save_if_best(
        model,
        val_acc,
        val_loss,
        macro_f1,
        global_epoch,
        "fine_tune"
    )

    if not improved:
        # save_if_best increments this counter when no improvement occurs.
        pass

    print("-" * 65)

    if epochs_without_improvement >= PATIENCE:
        print(
            f"Early stopping: validation accuracy has not improved "
            f"for {PATIENCE} consecutive epochs."
        )
        break


# ------------------------------------------------------------
# 9. Restore the best ConvNeXt checkpoint
# ------------------------------------------------------------

print("\nRestoring best ConvNeXt-Tiny checkpoint...")

checkpoint = torch.load(
    BEST_PATH,
    map_location="cpu",
    weights_only=True
)

model.load_state_dict(checkpoint["model_state_dict"])
model.to(device, memory_format=torch.channels_last)
model.eval()

print("\nConvNeXt-Tiny training complete.")
print(f"Best validation accuracy: {checkpoint['best_val_acc']:.2f}%")
print(f"Best epoch: {checkpoint['epoch']}")
print(f"Best phase: {checkpoint['phase']}")
print(f"Checkpoint: {BEST_PATH}")

history_df = pd.DataFrame(history_convnext)
display(history_df)


# ------------------------------------------------------------
# 10. Final per-class validation metrics
# ------------------------------------------------------------

val_loss, val_acc, macro_f1, y_true, y_pred = validate_model(model)

report = classification_report(
    y_true,
    y_pred,
    labels=list(range(NUM_CLASSES)),
    output_dict=True,
    zero_division=0
)

per_class_df = pd.DataFrame([
    {
        "class_id": class_id,
        "species": species_mapping.get(class_id, "Unknown"),
        "precision": report[str(class_id)]["precision"],
        "recall": report[str(class_id)]["recall"],
        "f1_score": report[str(class_id)]["f1-score"],
        "support": int(report[str(class_id)]["support"])
    }
    for class_id in range(NUM_CLASSES)
]).sort_values("f1_score")

print("\nFinal validation results")
print(f"Validation accuracy: {val_acc:.2f}%")
print(f"Macro-F1: {macro_f1:.4f}")
print("\n20 lowest-F1 species:")
display(per_class_df.head(20))

CSV_PATH = CHECKPOINT_DIR / "convnext_tiny_per_class_metrics.csv"
per_class_df.to_csv(CSV_PATH, index=False)

print(f"\nCheckpoint: {BEST_PATH}")
print(f"Per-class CSV: {CSV_PATH}")
print(
    "\nDownload the checkpoint and CSV before the Modal environment "
    "is terminated. /root is temporary storage."
)
```

```python

# ============================================================
# Inception V3: Auxiliary Classifier Fine-tuning
# Modal | NVIDIA L40S | PyTorch 2.8+
#
# Starts from:
# inception_v3_snakes_finetuned_v4_best.pth
#
# Adds auxiliary loss during training.
# Keeps original dataset and DataLoaders unchanged.
# ============================================================

import os
import gc
from pathlib import Path

import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd

from torchvision.models import inception_v3
from tqdm.auto import tqdm
from sklearn.metrics import f1_score, classification_report


# ------------------------------------------------------------
# 1. Configuration
# ------------------------------------------------------------

EPOCHS = 10
PATIENCE = 4

BACKBONE_LR = 5e-7
FINAL_BLOCK_LR = 1e-6
CLASSIFIER_LR = 1e-5
AUXILIARY_LR = 1e-5

AUX_LOSS_WEIGHT = 0.4
WEIGHT_DECAY = 1e-4

CHECKPOINT_DIR = Path("/root/snake_checkpoints")

SOURCE_PATH = (
    CHECKPOINT_DIR / "inception_v3_snakes_finetuned_v4_best.pth"
)

BEST_PATH = (
    CHECKPOINT_DIR / "inception_v3_snakes_aux_best.pth"
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

if device.type != "cuda":
    raise RuntimeError("CUDA GPU not detected.")

if not torch.cuda.is_bf16_supported():
    raise RuntimeError("BF16 is not supported by this GPU.")

torch.backends.cudnn.benchmark = True

print(f"GPU: {torch.cuda.get_device_name(0)}")
print(f"PyTorch: {torch.__version__}")
print(f"Source checkpoint: {SOURCE_PATH}")
print(f"Output checkpoint: {BEST_PATH}")


# ------------------------------------------------------------
# 2. Validate notebook variables and source checkpoint
# ------------------------------------------------------------

required_variables = [
    "train_loader",
    "val_loader",
    "class_to_idx",
    "idx_to_species",
    "NUM_CLASSES",
]

missing = [name for name in required_variables if name not in globals()]

if missing:
    raise RuntimeError(
        f"Missing notebook variables: {missing}. "
        "Run your existing dataset and DataLoader setup cells first."
    )

if not SOURCE_PATH.exists():
    raise FileNotFoundError(
        f"Source checkpoint not found: {SOURCE_PATH}. "
        "Restore your Inception V3 V4 checkpoint first."
    )

CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

class_mapping = {
    str(k): int(v) for k, v in class_to_idx.items()
}

species_mapping = {
    int(k): str(v) for k, v in idx_to_species.items()
}

source = torch.load(
    SOURCE_PATH,
    map_location="cpu",
    weights_only=True,
)

if int(source["num_classes"]) != int(NUM_CLASSES):
    raise ValueError("Checkpoint class count does not match NUM_CLASSES.")

if source["class_to_idx"] != class_mapping:
    raise ValueError("Checkpoint class mapping does not match the dataset.")

baseline_acc = float(source["best_val_acc"])

print(f"Baseline validation accuracy: {baseline_acc:.2f}%")


# ------------------------------------------------------------
# 3. Rebuild Inception V3 with auxiliary classifier enabled
# ------------------------------------------------------------

# Instantiate with aux_logits=True because the model needs its
# auxiliary branch during training.
model = inception_v3(
    weights=None,
    aux_logits=True,
    num_classes=1000,
    transform_input=False,
)

# Replace the main classification layer.
model.fc = nn.Linear(
    model.fc.in_features,
    NUM_CLASSES,
)

# Replace the auxiliary classifier output layer.
model.AuxLogits.fc = nn.Linear(
    model.AuxLogits.fc.in_features,
    NUM_CLASSES,
)

# Load existing main-network weights. The previous checkpoint
# was trained with the auxiliary classifier disabled, so it
# does not contain AuxLogits parameters.
source_state = source["model_state_dict"]

missing_keys, unexpected_keys = model.load_state_dict(
    source_state,
    strict=False,
)

# Only the auxiliary branch should be missing.
unexpected_aux = [
    key for key in unexpected_keys
    if not key.startswith("AuxLogits.")
]

missing_non_aux = [
    key for key in missing_keys
    if not key.startswith("AuxLogits.")
]

if unexpected_aux or missing_non_aux:
    raise RuntimeError(
        f"Unexpected checkpoint mismatch.\n"
        f"Unexpected keys: {unexpected_aux}\n"
        f"Missing non-auxiliary keys: {missing_non_aux}"
    )

del source, source_state
gc.collect()

model = model.to(
    device,
    memory_format=torch.channels_last,
)

print("Loaded existing Inception V3 weights.")
print("Initialized a new auxiliary classifier.")


# ------------------------------------------------------------
# 4. Freeze layers and select trainable blocks
# ------------------------------------------------------------

for parameter in model.parameters():
    parameter.requires_grad = False

# Intermediate blocks, final blocks, main classifier, and
# auxiliary classifier are trainable.
intermediate_modules = [
    model.Mixed_6a,
    model.Mixed_6b,
    model.Mixed_6c,
    model.Mixed_6d,
    model.Mixed_6e,
]

final_modules = [
    model.Mixed_7a,
    model.Mixed_7b,
    model.Mixed_7c,
]

for module in intermediate_modules + final_modules:
    for parameter in module.parameters():
        parameter.requires_grad = True

for module in [model.fc, model.AuxLogits]:
    for parameter in module.parameters():
        parameter.requires_grad = True

trainable_params = sum(
    p.numel() for p in model.parameters() if p.requires_grad
)

print(f"Trainable parameters: {trainable_params:,}")


# ------------------------------------------------------------
# 5. Optimizer and learning-rate scheduler
# ------------------------------------------------------------

optimizer = optim.AdamW(
    [
        {
            "params": [
                p for module in intermediate_modules
                for p in module.parameters()
                if p.requires_grad
            ],
            "lr": BACKBONE_LR,
        },
        {
            "params": [
                p for module in final_modules
                for p in module.parameters()
                if p.requires_grad
            ],
            "lr": FINAL_BLOCK_LR,
        },
        {
            "params": list(model.fc.parameters()),
            "lr": CLASSIFIER_LR,
        },
        {
            "params": list(model.AuxLogits.parameters()),
            "lr": AUXILIARY_LR,
        },
    ],
    weight_decay=WEIGHT_DECAY,
)

criterion = nn.CrossEntropyLoss()

scheduler = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=2,
    min_lr=1e-8,
)


# ------------------------------------------------------------
# 6. Verified checkpoint saving
# ------------------------------------------------------------

def save_checkpoint_safely(payload, destination):
    destination = Path(destination)
    destination.parent.mkdir(parents=True, exist_ok=True)

    temp_path = destination.with_name(
        destination.name + ".tmp"
    )

    try:
        safe_payload = dict(payload)

        safe_payload["class_to_idx"] = {
            str(k): int(v)
            for k, v in safe_payload["class_to_idx"].items()
        }

        safe_payload["idx_to_species"] = {
            int(k): str(v)
            for k, v in safe_payload["idx_to_species"].items()
        }

        safe_payload["num_classes"] = int(
            safe_payload["num_classes"]
        )

        safe_payload["best_val_acc"] = float(
            safe_payload["best_val_acc"]
        )

        safe_payload["epoch"] = int(
            safe_payload["epoch"]
        )

        safe_payload["model_state_dict"] = {
            k: v.detach().cpu().contiguous().clone()
            for k, v in safe_payload["model_state_dict"].items()
        }

        with open(temp_path, "wb") as f:
            torch.save(safe_payload, f)
            f.flush()
            os.fsync(f.fileno())

        verified = torch.load(
            temp_path,
            map_location="cpu",
            weights_only=True,
        )

        assert "model_state_dict" in verified
        assert len(verified["model_state_dict"]) > 0
        assert "best_val_acc" in verified
        assert "class_to_idx" in verified

        os.replace(temp_path, destination)

        print(
            f"Verified checkpoint saved: {destination} "
            f"({destination.stat().st_size / 1e6:.1f} MB)"
        )

    except Exception:
        if temp_path.exists():
            temp_path.unlink()
        raise


# ------------------------------------------------------------
# 7. Validation
# ------------------------------------------------------------

def validate_model():
    model.eval()

    total_loss = 0.0
    total_correct = 0
    total_samples = 0

    y_true = []
    y_pred = []

    with torch.inference_mode():
        for images, labels in tqdm(
            val_loader,
            desc="Validation",
            leave=False,
        ):
            images = images.to(
                device,
                non_blocking=True,
                memory_format=torch.channels_last,
            )

            labels = labels.to(
                device,
                non_blocking=True,
            ).long()

            with torch.autocast(
                device_type="cuda",
                dtype=torch.bfloat16,
            ):
                outputs = model(images)
                # In evaluation mode, torchvision Inception V3
                # returns main logits, not the auxiliary output.
                loss = criterion(outputs, labels)

            predictions = outputs.argmax(dim=1)
            batch_size = labels.size(0)

            total_loss += loss.item() * batch_size
            total_correct += (
                predictions == labels
            ).sum().item()
            total_samples += batch_size

            y_true.extend(labels.cpu().tolist())
            y_pred.extend(predictions.cpu().tolist())

    val_loss = total_loss / total_samples
    val_acc = 100.0 * total_correct / total_samples

    macro_f1 = f1_score(
        y_true,
        y_pred,
        labels=list(range(NUM_CLASSES)),
        average="macro",
        zero_division=0,
    )

    return val_loss, val_acc, macro_f1, y_true, y_pred


# ------------------------------------------------------------
# 8. Preserve baseline checkpoint before training
# ------------------------------------------------------------

# The new auxiliary branch is randomly initialized, so the
# starting model is not identical to the old inference model.
# Save its state as a recovery point, but don't count it as
# a validated improvement.
initial_payload = {
    "model_state_dict": model.state_dict(),
    "class_to_idx": class_mapping,
    "idx_to_species": species_mapping,
    "num_classes": NUM_CLASSES,
    "best_val_acc": baseline_acc,
    "epoch": 0,
    "architecture": "inception_v3",
    "phase": "auxiliary_finetuning",
    "aux_loss_weight": AUX_LOSS_WEIGHT,
    "baseline_val_acc": baseline_acc,
}

save_checkpoint_safely(initial_payload, BEST_PATH)

del initial_payload
gc.collect()

best_val_acc = baseline_acc
best_epoch = 0
epochs_without_improvement = 0
history_aux = []


# ------------------------------------------------------------
# 9. Training loop with auxiliary loss
# ------------------------------------------------------------

for epoch in range(EPOCHS):

    # IMPORTANT: set the ROOT model to training mode. Torchvision's
    # Inception3.forward() checks model.training before returning aux logits.
    # Calling .train() only on child modules leaves model.training=False.
    model.train()

    # Keep BatchNorm running statistics fixed during fine-tuning.
    # This does not change model.training, so the auxiliary branch still runs.
    for module in model.modules():
        if isinstance(module, nn.modules.batchnorm._BatchNorm):
            module.eval()

    train_loss_sum = 0.0
    main_loss_sum = 0.0
    aux_loss_sum = 0.0

    train_correct = 0
    train_total = 0

    for images, labels in tqdm(
        train_loader,
        desc=f"Epoch {epoch + 1}/{EPOCHS} [Train]",
    ):
        images = images.to(
            device,
            non_blocking=True,
            memory_format=torch.channels_last,
        )

        labels = labels.to(
            device,
            non_blocking=True,
        ).long()

        optimizer.zero_grad(set_to_none=True)

        
        with torch.autocast(
            device_type="cuda",
            dtype=torch.bfloat16,
        ):
            outputs = model(images)
        
            if isinstance(outputs, tuple):
                main_logits, aux_logits = outputs
            elif hasattr(outputs, "logits") and hasattr(outputs, "aux_logits"):
                main_logits = outputs.logits
                aux_logits = outputs.aux_logits
            else:
                raise RuntimeError(
                    "Expected Inception auxiliary outputs during training, but "
                    f"received {type(outputs).__name__}. "
                    f"model.training={model.training}, "
                    f"model.aux_logits={model.aux_logits}, "
                    f"AuxLogits exists={model.AuxLogits is not None}. "
                    "Ensure model.train() is called on the root model before "
                    "the training forward pass."
                )

            if aux_logits is None:
                raise RuntimeError(
                    "Inception returned aux_logits=None. The root model must be "
                    "in training mode (model.training=True) for the auxiliary "
                    "classifier to produce outputs."
                )

            main_loss = criterion(main_logits, labels)
            aux_loss = criterion(aux_logits, labels)
        
            loss = main_loss + AUX_LOSS_WEIGHT * aux_loss


        loss.backward()
        optimizer.step()

        batch_size = labels.size(0)

        train_loss_sum += loss.item() * batch_size
        main_loss_sum += main_loss.item() * batch_size
        aux_loss_sum += aux_loss.item() * batch_size

        train_correct += (
            main_logits.argmax(dim=1) == labels
        ).sum().item()

        train_total += batch_size

    train_loss = train_loss_sum / train_total
    train_main_loss = main_loss_sum / train_total
    train_aux_loss = aux_loss_sum / train_total
    train_acc = 100.0 * train_correct / train_total

    # --------------------------------------------------------
    # Validation uses the main classifier only
    # --------------------------------------------------------

    val_loss, val_acc, macro_f1, _, _ = validate_model()

    scheduler.step(val_loss)

    print(f"\nEpoch {epoch + 1}/{EPOCHS}")
    print(f"Combined train loss: {train_loss:.4f}")
    print(f"Main train loss: {train_main_loss:.4f}")
    print(f"Auxiliary train loss: {train_aux_loss:.4f}")
    print(f"Train accuracy: {train_acc:.2f}%")
    print(f"Validation loss: {val_loss:.4f}")
    print(f"Validation accuracy: {val_acc:.2f}%")
    print(f"Macro-F1: {macro_f1:.4f}")

    print(
        "Learning rates: "
        f"Mixed6={optimizer.param_groups[0]['lr']:.2e}, "
        f"Mixed7={optimizer.param_groups[1]['lr']:.2e}, "
        f"FC={optimizer.param_groups[2]['lr']:.2e}, "
        f"Aux={optimizer.param_groups[3]['lr']:.2e}"
    )

    history_aux.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_main_loss": train_main_loss,
        "train_aux_loss": train_aux_loss,
        "train_acc": train_acc,
        "val_loss": val_loss,
        "val_acc": val_acc,
        "macro_f1": macro_f1,
    })

    # Save only when the main classifier beats the previous
    # best validated Inception model.
    if val_acc > best_val_acc:
        payload = {
            "model_state_dict": model.state_dict(),
            "class_to_idx": class_mapping,
            "idx_to_species": species_mapping,
            "num_classes": NUM_CLASSES,
            "best_val_acc": float(val_acc),
            "val_loss": float(val_loss),
            "val_macro_f1": float(macro_f1),
            "epoch": epoch + 1,
            "architecture": "inception_v3",
            "phase": "auxiliary_finetuning",
            "aux_loss_weight": AUX_LOSS_WEIGHT,
            "baseline_val_acc": baseline_acc,
        }

        save_checkpoint_safely(payload, BEST_PATH)

        best_val_acc = float(val_acc)
        best_epoch = epoch + 1
        epochs_without_improvement = 0

        print(
            f"New best validation accuracy: {best_val_acc:.2f}%"
        )

        del payload

    else:
        epochs_without_improvement += 1
        print(f"Best accuracy retained: {best_val_acc:.2f}%")

    print("-" * 70)

    if epochs_without_improvement >= PATIENCE:
        print("Early stopping: validation accuracy has stalled.")
        break


# ------------------------------------------------------------
# 10. Restore best model and report results
# ------------------------------------------------------------

best_checkpoint = torch.load(
    BEST_PATH,
    map_location="cpu",
    weights_only=True,
)

model.load_state_dict(best_checkpoint["model_state_dict"])
model.to(device, memory_format=torch.channels_last)
model.eval()

print("\nAuxiliary-classifier experiment complete.")
print(f"Previous baseline: {baseline_acc:.2f}%")
print(f"Best retained accuracy: {best_checkpoint['best_val_acc']:.2f}%")
print(f"Checkpoint epoch: {best_checkpoint['epoch']}")
print(f"Checkpoint: {BEST_PATH}")

history_df_aux = pd.DataFrame(history_aux)
display(history_df_aux)


# ------------------------------------------------------------
# 11. Final per-class validation metrics
# ------------------------------------------------------------

_, final_acc, final_f1, y_true, y_pred = validate_model()

report = classification_report(
    y_true,
    y_pred,
    labels=list(range(NUM_CLASSES)),
    output_dict=True,
    zero_division=0,
)

per_class_df_aux = pd.DataFrame([
    {
        "class_id": class_id,
        "species": species_mapping.get(class_id, "Unknown"),
        "precision": report[str(class_id)]["precision"],
        "recall": report[str(class_id)]["recall"],
        "f1_score": report[str(class_id)]["f1-score"],
        "support": int(report[str(class_id)]["support"]),
    }
    for class_id in range(NUM_CLASSES)
]).sort_values("f1_score")

print(f"\nFinal validation accuracy: {final_acc:.2f}%")
print(f"Final macro-F1: {final_f1:.4f}")
print("\n20 lowest-F1 species:")
display(per_class_df_aux.head(20))

CSV_PATH = CHECKPOINT_DIR / "inception_v3_aux_per_class_metrics.csv"
per_class_df_aux.to_csv(CSV_PATH, index=False)

print(f"\nPer-class metrics: {CSV_PATH}")
print(
    "\nDownload the checkpoint and CSV before the Modal environment "
    "is terminated. /root is temporary storage."
)
```
